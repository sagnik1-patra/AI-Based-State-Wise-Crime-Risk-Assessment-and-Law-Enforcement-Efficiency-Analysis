```python
import os
import json
import random
import warnings
from pathlib import Path
from datetime import datetime

import numpy as np
import pandas as pd
import yaml
import joblib
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.optimizers import Adam

warnings.filterwarnings("ignore")

# ============================================================
# 1. PATH CONFIGURATION
# ============================================================

PROJECT_DIR = Path(
    r"C:\Users\sagni\Downloads"
    r"\AI-Based State-Wise Crime Risk Assessment "
    r"and Law Enforcement Efficiency Analysis"
)

DATASET_PATH = PROJECT_DIR / "NCRB_Table_1A.1.csv"

PROJECT_DIR.mkdir(parents=True, exist_ok=True)

H5_PATH = PROJECT_DIR / "crime_risk_model.h5"
PKL_PATH = PROJECT_DIR / "crime_risk_model.pkl"
YAML_PATH = PROJECT_DIR / "config.yaml"
JSON_PATH = PROJECT_DIR / "model_info.json"

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

print("=" * 65)
print("AI-BASED STATE-WISE CRIME RISK ASSESSMENT")
print("=" * 65)

# ============================================================
# 2. LOAD DATASET
# ============================================================

if not DATASET_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found:\n{DATASET_PATH}"
    )

df = pd.read_csv(DATASET_PATH)

print("\nOriginal Dataset Shape:", df.shape)
print("\nDataset Columns:")
print(df.columns.tolist())

print("\nFirst Five Records:")
print(df.head())

# ============================================================
# 3. DATA CLEANING
# ============================================================

required_columns = [
    "State/UT",
    "2020",
    "2021",
    "2022",
    "Mid-Year Projected Population (in Lakhs) (2022)",
    "Rate of Cognizable Crimes (IPC) (2022)",
    "Chargesheeting Rate (2022)"
]

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing columns: {missing_columns}"
    )

# Exclude aggregate rows.
# Only individual states and UTs should be modeled.

df["State/UT"] = df["State/UT"].astype(str).str.strip()

df = df[
    ~df["State/UT"].str.lower().str.startswith("total")
].copy()

numeric_columns = required_columns[1:]

for col in numeric_columns:
    df[col] = pd.to_numeric(
        df[col].astype(str).str.replace(",", "", regex=False),
        errors="coerce"
    )

# Rows without a valid target cannot be used.

target_rate_col = "Rate of Cognizable Crimes (IPC) (2022)"

df = df.dropna(subset=[target_rate_col])
df = df[df[target_rate_col] >= 0].copy()

df = df.reset_index(drop=True)

print("\nCleaned Dataset Shape:", df.shape)
print("\nMissing Values:")
print(df.isnull().sum())

# ============================================================
# 4. FEATURE ENGINEERING
# ============================================================

# Historical counts are used as model inputs.
# Do not use 2022 crimes, 2022 crime rate,
# or 2022 chargesheeting rate as predictors.

df["Crime_Growth_2020_2021"] = (
    (df["2021"] - df["2020"])
    / df["2020"].replace(0, np.nan)
) * 100

df["Crime_Growth_2020_2021"] = (
    df["Crime_Growth_2020_2021"]
    .replace([np.inf, -np.inf], np.nan)
)

df["Historical_Average_Crimes"] = (
    df[["2020", "2021"]].mean(axis=1)
)

df["Historical_Crime_Difference"] = (
    df["2021"] - df["2020"]
)

# Population is from 2022, so it is not used
# as an input for a strictly historical forecast.

feature_columns = [
    "2020",
    "2021",
    "Crime_Growth_2020_2021",
    "Historical_Average_Crimes",
    "Historical_Crime_Difference"
]

# ============================================================
# 5. CREATE CRIME RISK CATEGORIES
# ============================================================

# Relative risk categories are defined using
# the distribution of reported 2022 crime rates.
#
# These are dataset-specific categories,
# not official NCRB risk classifications.

q1 = float(df[target_rate_col].quantile(1 / 3))
q2 = float(df[target_rate_col].quantile(2 / 3))

if q1 == q2:
    raise ValueError(
        "Cannot construct three risk categories: "
        "crime-rate quantiles are identical."
    )

def classify_crime_risk(rate):
    if rate <= q1:
        return "Low Risk"
    elif rate <= q2:
        return "Medium Risk"
    else:
        return "High Risk"

df["Crime_Risk"] = (
    df[target_rate_col].apply(classify_crime_risk)
)

print("\nCrime Risk Distribution:")
print(df["Crime_Risk"].value_counts())

print("\nRisk Thresholds:")
print(f"Low Risk: <= {q1:.2f}")
print(f"Medium Risk: > {q1:.2f} and <= {q2:.2f}")
print(f"High Risk: > {q2:.2f}")

# ============================================================
# 6. LAW ENFORCEMENT EFFICIENCY ANALYSIS
# ============================================================

chargesheet_col = "Chargesheeting Rate (2022)"

print("\nChargesheeting Rate Statistics:")
print(df[chargesheet_col].describe())

# Chargesheeting is a descriptive indicator only.
# It is not a complete measure of police effectiveness.

df["Chargesheeting_Category"] = pd.cut(
    df[chargesheet_col],
    bins=[-np.inf, 50, 75, np.inf],
    labels=[
        "Lower Chargesheeting",
        "Moderate Chargesheeting",
        "Higher Chargesheeting"
    ]
)

print("\nChargesheeting Categories:")
print(df["Chargesheeting_Category"].value_counts())

# ============================================================
# 7. PREPARE FEATURES AND TARGET
# ============================================================

X = df[feature_columns].copy()

label_encoder = LabelEncoder()
y = label_encoder.fit_transform(df["Crime_Risk"])

class_names = label_encoder.classes_.tolist()
num_classes = len(class_names)

if num_classes != 3:
    raise ValueError(
        "Expected three crime-risk classes."
    )

print("\nEncoded Classes:")
for idx, name in enumerate(class_names):
    print(idx, ":", name)

# ============================================================
# 8. TRAIN-TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=SEED,
    stratify=y
)

print("\nTraining Samples:", len(X_train))
print("Testing Samples:", len(X_test))

# ============================================================
# 9. RANDOM FOREST MODEL
# ============================================================

rf_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="median")
    ),
    (
        "scaler",
        StandardScaler()
    ),
    (
        "classifier",
        RandomForestClassifier(
            n_estimators=200,
            max_depth=5,
            min_samples_leaf=2,
            random_state=SEED,
            class_weight="balanced"
        )
    )
])

rf_pipeline.fit(X_train, y_train)

rf_predictions = rf_pipeline.predict(X_test)

rf_accuracy = accuracy_score(y_test, rf_predictions)
rf_precision = precision_score(
    y_test, rf_predictions,
    average="weighted", zero_division=0
)
rf_recall = recall_score(
    y_test, rf_predictions,
    average="weighted", zero_division=0
)
rf_f1 = f1_score(
    y_test, rf_predictions,
    average="weighted", zero_division=0
)

print("\n" + "=" * 65)
print("RANDOM FOREST RESULTS")
print("=" * 65)

print(f"Accuracy : {rf_accuracy:.4f}")
print(f"Precision: {rf_precision:.4f}")
print(f"Recall   : {rf_recall:.4f}")
print(f"F1 Score : {rf_f1:.4f}")

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        rf_predictions,
        labels=np.arange(num_classes),
        target_names=class_names,
        zero_division=0
    )
)

# ============================================================
# 10. SAVE RANDOM FOREST MODEL AS PKL
# ============================================================

model_bundle = {
    "model": rf_pipeline,
    "label_encoder": label_encoder,
    "feature_columns": feature_columns,
    "risk_thresholds": {
        "low_max": q1,
        "medium_max": q2
    },
    "class_names": class_names
}

joblib.dump(model_bundle, PKL_PATH)

print("\nPKL Model Saved:", PKL_PATH)

# ============================================================
# 11. PREPROCESS DATA FOR NEURAL NETWORK
# ============================================================

nn_imputer = SimpleImputer(strategy="median")
nn_scaler = StandardScaler()

X_train_nn = nn_imputer.fit_transform(X_train)
X_test_nn = nn_imputer.transform(X_test)

X_train_nn = nn_scaler.fit_transform(X_train_nn)
X_test_nn = nn_scaler.transform(X_test_nn)

# ============================================================
# 12. BUILD NEURAL NETWORK
# ============================================================

nn_model = Sequential([
    Input(shape=(X_train_nn.shape[1],)),

    Dense(
        32,
        activation="relu"
    ),

    Dropout(0.2),

    Dense(
        16,
        activation="relu"
    ),

    Dense(
        num_classes,
        activation="softmax"
    )
])

nn_model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print("\nNeural Network Architecture:")
nn_model.summary()

# ============================================================
# 13. TRAIN NEURAL NETWORK
# ============================================================

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=15,
    restore_best_weights=True
)

history = nn_model.fit(
    X_train_nn,
    y_train,
    epochs=150,
    batch_size=8,
    validation_split=0.2,
    callbacks=[early_stopping],
    verbose=0,
    shuffle=True
)

print("\nNeural Network Training Completed")
print("Epochs Used:", len(history.history["loss"]))

# ============================================================
# 14. EVALUATE NEURAL NETWORK
# ============================================================

nn_probabilities = nn_model.predict(
    X_test_nn,
    verbose=0
)

nn_predictions = np.argmax(
    nn_probabilities,
    axis=1
)

nn_accuracy = accuracy_score(
    y_test,
    nn_predictions
)

nn_precision = precision_score(
    y_test,
    nn_predictions,
    average="weighted",
    zero_division=0
)

nn_recall = recall_score(
    y_test,
    nn_predictions,
    average="weighted",
    zero_division=0
)

nn_f1 = f1_score(
    y_test,
    nn_predictions,
    average="weighted",
    zero_division=0
)

print("\n" + "=" * 65)
print("NEURAL NETWORK RESULTS")
print("=" * 65)

print(f"Accuracy : {nn_accuracy:.4f}")
print(f"Precision: {nn_precision:.4f}")
print(f"Recall   : {nn_recall:.4f}")
print(f"F1 Score : {nn_f1:.4f}")

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        nn_predictions,
        labels=np.arange(num_classes),
        target_names=class_names,
        zero_division=0
    )
)

# ============================================================
# 15. SAVE NEURAL NETWORK AS H5
# ============================================================

nn_model.save(str(H5_PATH))

print("\nH5 Model Saved:", H5_PATH)

# The H5 file contains the neural network.
# Save its fitted preprocessing objects separately
# so future predictions use the same transformations.

NN_PREPROCESSOR_PATH = (
    PROJECT_DIR / "nn_preprocessor.pkl"
)

joblib.dump(
    {
        "imputer": nn_imputer,
        "scaler": nn_scaler,
        "label_encoder": label_encoder,
        "feature_columns": feature_columns,
        "risk_thresholds": {
            "low_max": q1,
            "medium_max": q2
        }
    },
    NN_PREPROCESSOR_PATH
)

print("NN Preprocessor Saved:", NN_PREPROCESSOR_PATH)

# ============================================================
# 16. MODEL COMPARISON
# ============================================================

comparison_df = pd.DataFrame({
    "Model": [
        "Random Forest",
        "Neural Network"
    ],
    "Accuracy": [
        rf_accuracy,
        nn_accuracy
    ],
    "Precision": [
        rf_precision,
        nn_precision
    ],
    "Recall": [
        rf_recall,
        nn_recall
    ],
    "F1 Score": [
        rf_f1,
        nn_f1
    ]
})

print("\nModel Comparison:")
print(comparison_df.to_string(index=False))

# ============================================================
# 17. CREATE CONFIG.YAML
# ============================================================

config = {
    "project": {
        "name": (
            "AI-Based State-Wise Crime Risk Assessment "
            "and Law Enforcement Efficiency Analysis"
        ),
        "dataset": str(DATASET_PATH),
        "output_directory": str(PROJECT_DIR),
        "random_seed": SEED
    },

    "dataset": {
        "records_used": int(len(df)),
        "features": feature_columns,
        "target": "Crime_Risk",
        "target_source": target_rate_col,
        "classes": class_names,
        "excluded_aggregate_rows": True
    },

    "risk_classification": {
        "method": "2022 crime rate tertiles",
        "low_max": q1,
        "medium_max": q2,
        "note": (
            "Relative dataset categories, "
            "not official NCRB risk labels"
        )
    },

    "training": {
        "test_size": 0.25,
        "random_state": SEED,
        "stratified_split": True
    },

    "random_forest": {
        "n_estimators": 200,
        "max_depth": 5,
        "min_samples_leaf": 2,
        "class_weight": "balanced"
    },

    "neural_network": {
        "hidden_layers": [32, 16],
        "dropout": 0.2,
        "learning_rate": 0.001,
        "epochs_max": 150,
        "batch_size": 8,
        "early_stopping_patience": 15
    },

    "saved_models": {
        "random_forest": PKL_PATH.name,
        "neural_network": H5_PATH.name,
        "neural_network_preprocessor": (
            NN_PREPROCESSOR_PATH.name
        )
    }
}

with open(YAML_PATH, "w", encoding="utf-8") as f:
    yaml.safe_dump(
        config,
        f,
        sort_keys=False,
        allow_unicode=True
    )

print("\nYAML Configuration Saved:", YAML_PATH)

# ============================================================
# 18. CREATE MODEL_INFO.JSON
# ============================================================

model_info = {
    "project_name": (
        "AI-Based State-Wise Crime Risk Assessment "
        "and Law Enforcement Efficiency Analysis"
    ),

    "created_at": datetime.now().isoformat(),

    "dataset": {
        "original_rows": 39,
        "processed_rows": int(len(df)),
        "training_rows": int(len(X_train)),
        "testing_rows": int(len(X_test)),
        "number_of_features": len(feature_columns),
        "feature_names": feature_columns,
        "classes": class_names
    },

    "risk_thresholds": {
        "low_max": q1,
        "medium_max": q2
    },

    "random_forest": {
        "accuracy": float(rf_accuracy),
        "precision": float(rf_precision),
        "recall": float(rf_recall),
        "f1_score": float(rf_f1),
        "confusion_matrix": confusion_matrix(
            y_test,
            rf_predictions,
            labels=np.arange(num_classes)
        ).tolist()
    },

    "neural_network": {
        "accuracy": float(nn_accuracy),
        "precision": float(nn_precision),
        "recall": float(nn_recall),
        "f1_score": float(nn_f1),
        "epochs_trained": int(
            len(history.history["loss"])
        ),
        "confusion_matrix": confusion_matrix(
            y_test,
            nn_predictions,
            labels=np.arange(num_classes)
        ).tolist()
    },

    "chargesheeting_analysis": {
        "mean": float(df[chargesheet_col].mean()),
        "median": float(df[chargesheet_col].median()),
        "minimum": float(df[chargesheet_col].min()),
        "maximum": float(df[chargesheet_col].max())
    },

    "limitations": [
        "Small state-level dataset",
        "Risk labels are derived from 2022 crime rates",
        "2022 population is excluded from predictors",
        "Single train-test split has high uncertainty",
        "Chargesheeting rate is descriptive only",
        "Not validated for operational crime forecasting"
    ],

    "saved_files": {
        "h5": H5_PATH.name,
        "pkl": PKL_PATH.name,
        "yaml": YAML_PATH.name,
        "json": JSON_PATH.name,
        "nn_preprocessor": NN_PREPROCESSOR_PATH.name
    }
}

with open(JSON_PATH, "w", encoding="utf-8") as f:
    json.dump(
        model_info,
        f,
        indent=4,
        ensure_ascii=False
    )

print("\nJSON Model Information Saved:", JSON_PATH)

# ============================================================
# 19. VERIFY SAVED FILES
# ============================================================

print("\n" + "=" * 65)
print("VERIFYING SAVED ARTIFACTS")
print("=" * 65)

saved_files = [
    H5_PATH,
    PKL_PATH,
    YAML_PATH,
    JSON_PATH,
    NN_PREPROCESSOR_PATH
]

for file_path in saved_files:
    if file_path.exists():
        size_kb = file_path.stat().st_size / 1024
        print(
            f"[SUCCESS] {file_path.name} "
            f"({size_kb:.2f} KB)"
        )
    else:
        print(f"[FAILED] {file_path.name}")

# ============================================================
# 20. TEST LOADING SAVED MODELS
# ============================================================

loaded_bundle = joblib.load(PKL_PATH)
loaded_rf = loaded_bundle["model"]

rf_test_prediction = loaded_rf.predict(
    X_test.iloc[[0]]
)

print("\nLoaded Random Forest Prediction:")
print(
    label_encoder.inverse_transform(
        rf_test_prediction
    )[0]
)

loaded_nn = tf.keras.models.load_model(
    str(H5_PATH),
    compile=False
)

loaded_preprocessor = joblib.load(
    NN_PREPROCESSOR_PATH
)

sample = X_test.iloc[[0]]

sample_processed = (
    loaded_preprocessor["imputer"].transform(sample)
)

sample_processed = (
    loaded_preprocessor["scaler"].transform(
        sample_processed
    )
)

nn_test_prediction = np.argmax(
    loaded_nn.predict(
        sample_processed,
        verbose=0
    ),
    axis=1
)

print("\nLoaded Neural Network Prediction:")
print(
    label_encoder.inverse_transform(
        nn_test_prediction
    )[0]
)

# ============================================================
# 21. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 65)
print("PROJECT COMPLETED SUCCESSFULLY")
print("=" * 65)

print("\nOutput Directory:")
print(PROJECT_DIR)

print("\nRandom Forest Accuracy:", round(rf_accuracy, 4))
print("Neural Network Accuracy:", round(nn_accuracy, 4))

print("\nGenerated Files:")
for file_path in saved_files:
    print(" -", file_path.name)

print("\nNOTE:")
print(
    "Results are exploratory due to the small "
    "dataset and should not be interpreted as "
)
```
