```python
import json
import random
import warnings
from pathlib import Path
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import yaml

from matplotlib.patches import Patch

from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    cross_val_score
)
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    classification_report,
    confusion_matrix
)

warnings.filterwarnings("ignore")

# ================================================================
# 1. PROJECT CONFIGURATION
# ================================================================

PROJECT_DIR = Path(
    r"C:\Users\sagni\Downloads"
    r"\AI-Based State-Wise Crime Risk Assessment "
    r"and Law Enforcement Efficiency Analysis"
)

DATASET_PATH = PROJECT_DIR / "NCRB_Table_1A.1.csv"
BASELINE_PATH = PROJECT_DIR / "crime_risk_model.pkl"
AIS_PATH = PROJECT_DIR / "ais_crime_risk_model.pkl"

PROJECT_DIR.mkdir(parents=True, exist_ok=True)

SEED = 42
random.seed(SEED)
np.random.seed(SEED)

rng = np.random.default_rng(SEED)

sns.set_theme(style="whitegrid")

def output(filename):
    return PROJECT_DIR / ("pso_" + filename)

def save_plot(filename):
    plt.tight_layout()
    plt.savefig(
        output(filename),
        dpi=300,
        bbox_inches="tight"
    )
    plt.show()
    plt.close()
    print("Saved:", output(filename).name)

print("=" * 75)
print("PSO-BASED CRIME RISK ASSESSMENT")
print("=" * 75)

# ================================================================
# 2. LOAD DATASET
# ================================================================

if not DATASET_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found:\n{DATASET_PATH}"
    )

df = pd.read_csv(DATASET_PATH)

print("\nOriginal dataset:", df.shape)
print("\nColumns:", df.columns.tolist())

df["State/UT"] = (
    df["State/UT"].astype(str).str.strip()
)

# Remove aggregate rows

df = df[
    ~df["State/UT"]
    .str.lower()
    .str.startswith("total")
].copy()

numeric_columns = [
    "2020",
    "2021",
    "2022",
    "Mid-Year Projected Population (in Lakhs) (2022)",
    "Rate of Cognizable Crimes (IPC) (2022)",
    "Chargesheeting Rate (2022)"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(
        df[col].astype(str).str.replace(
            ",", "", regex=False
        ),
        errors="coerce"
    )

RATE_COL = "Rate of Cognizable Crimes (IPC) (2022)"
CHARGE_COL = "Chargesheeting Rate (2022)"

df = df.dropna(subset=[RATE_COL])
df = df[df[RATE_COL] >= 0]
df = df.reset_index(drop=True)

print("\nCleaned dataset:", df.shape)

# ================================================================
# 3. FEATURE ENGINEERING
# ================================================================

# Historical inputs only.
# Do not include 2022 outcomes as predictors.

df["Crime_Growth_2020_2021"] = (
    (df["2021"] - df["2020"])
    / df["2020"].replace(0, np.nan)
) * 100

df["Crime_Growth_2020_2021"] = (
    df["Crime_Growth_2020_2021"]
    .replace([np.inf, -np.inf], np.nan)
)

df["Historical_Average_Crimes"] = (
    df[["2020", "2021"]].mean(axis=1)
)

df["Historical_Crime_Difference"] = (
    df["2021"] - df["2020"]
)

feature_columns = [
    "2020",
    "2021",
    "Crime_Growth_2020_2021",
    "Historical_Average_Crimes",
    "Historical_Crime_Difference"
]

# ================================================================
# 4. RISK LABELS
# ================================================================

baseline_bundle = None

if BASELINE_PATH.exists():

    baseline_bundle = joblib.load(BASELINE_PATH)

    thresholds = baseline_bundle["risk_thresholds"]

    q1 = float(thresholds["low_max"])
    q2 = float(thresholds["medium_max"])

else:

    q1 = float(df[RATE_COL].quantile(1 / 3))
    q2 = float(df[RATE_COL].quantile(2 / 3))

if q1 >= q2:
    raise ValueError("Invalid risk thresholds.")

def classify_risk(rate):

    if rate <= q1:
        return "Low Risk"

    if rate <= q2:
        return "Medium Risk"

    return "High Risk"

df["Crime_Risk"] = df[RATE_COL].apply(classify_risk)

encoder = LabelEncoder()

X = df[feature_columns].copy()
y = encoder.fit_transform(df["Crime_Risk"])

class_names = encoder.classes_.tolist()
n_classes = len(class_names)
n_features = len(feature_columns)

print("\nRisk classes:", class_names)
print("\nRisk distribution:")
print(df["Crime_Risk"].value_counts())

# ================================================================
# 5. TRAIN-TEST SPLIT
# ================================================================

indices = np.arange(len(df))

train_idx, test_idx = train_test_split(
    indices,
    test_size=0.25,
    random_state=SEED,
    stratify=y
)

X_train = X.iloc[train_idx].copy()
X_test = X.iloc[test_idx].copy()

y_train = y[train_idx]
y_test = y[test_idx]

print("\nTraining records:", len(X_train))
print("Testing records:", len(X_test))

# ================================================================
# 6. PSO PARTICLE REPRESENTATION
# ================================================================

# Particle dimensions:
#
# 0: n_estimators
# 1: max_depth
# 2: min_samples_split
# 3: min_samples_leaf
# 4: max_features
# 5+: feature selection scores
#
# Feature scores >= 0.5 mean selected.

lower_bounds = np.array(
    [50, 2, 2, 1, 0.4] + [0] * n_features,
    dtype=float
)

upper_bounds = np.array(
    [300, 12, 10, 5, 1.0] + [1] * n_features,
    dtype=float
)

dimensions = len(lower_bounds)

def decode_particle(position):

    position = np.clip(
        position,
        lower_bounds,
        upper_bounds
    )

    params = {
        "n_estimators": int(round(position[0])),
        "max_depth": int(round(position[1])),
        "min_samples_split": int(round(position[2])),
        "min_samples_leaf": int(round(position[3])),
        "max_features": float(position[4])
    }

    mask = position[5:] >= 0.5

    if not np.any(mask):
        mask[np.argmax(position[5:])] = True

    selected_features = [
        feature_columns[i]
        for i in range(n_features)
        if mask[i]
    ]

    return params, selected_features

# ================================================================
# 7. RANDOM FOREST PIPELINE
# ================================================================

def build_pipeline(params):

    return Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        ),
        (
            "classifier",
            RandomForestClassifier(
                **params,
                random_state=SEED,
                class_weight="balanced",
                n_jobs=1
            )
        )
    ])

# ================================================================
# 8. PSO FITNESS FUNCTION
# ================================================================

cv = StratifiedKFold(
    n_splits=3,
    shuffle=True,
    random_state=SEED
)

fitness_cache = {}

def fitness(position):

    params, selected_features = decode_particle(
        position
    )

    key = (
        params["n_estimators"],
        params["max_depth"],
        params["min_samples_split"],
        params["min_samples_leaf"],
        round(params["max_features"], 3),
        tuple(selected_features)
    )

    if key in fitness_cache:
        return fitness_cache[key]

    model = build_pipeline(params)

    scores = cross_val_score(
        model,
        X_train[selected_features],
        y_train,
        cv=cv,
        scoring="f1_macro",
        n_jobs=1,
        error_score="raise"
    )

    mean_f1 = float(np.mean(scores))

    feature_penalty = (
        0.005 * len(selected_features) / n_features
    )

    # PSO minimizes fitness.
    # Negative macro F1 converts maximization
    # into minimization.

    cost = -(mean_f1 - feature_penalty)

    fitness_cache[key] = cost

    return cost

# ================================================================
# 9. PSO INITIALIZATION
# ================================================================

N_PARTICLES = 12
ITERATIONS = 12

INERTIA = 0.7
COGNITIVE = 1.5
SOCIAL = 1.5

positions = rng.uniform(
    lower_bounds,
    upper_bounds,
    size=(N_PARTICLES, dimensions)
)

velocities = np.zeros_like(positions)

personal_best_positions = positions.copy()

personal_best_costs = np.array([
    fitness(p) for p in positions
])

best_idx = np.argmin(personal_best_costs)

global_best_position = (
    personal_best_positions[best_idx].copy()
)

global_best_cost = float(
    personal_best_costs[best_idx]
)

optimization_history = []

print("\nStarting PSO optimization...")

# ================================================================
# 10. PSO OPTIMIZATION LOOP
# ================================================================

for iteration in range(ITERATIONS):

    current_costs = []

    for i in range(N_PARTICLES):

        r1 = rng.random(dimensions)
        r2 = rng.random(dimensions)

        cognitive_velocity = (
            COGNITIVE
            * r1
            * (
                personal_best_positions[i]
                - positions[i]
            )
        )

        social_velocity = (
            SOCIAL
            * r2
            * (
                global_best_position
                - positions[i]
            )
        )

        velocities[i] = (
            INERTIA * velocities[i]
            + cognitive_velocity
            + social_velocity
        )

        # Limit particle velocity.

        max_velocity = (
            upper_bounds - lower_bounds
        ) * 0.25

        velocities[i] = np.clip(
            velocities[i],
            -max_velocity,
            max_velocity
        )

        positions[i] += velocities[i]

        positions[i] = np.clip(
            positions[i],
            lower_bounds,
            upper_bounds
        )

        current_cost = fitness(positions[i])

        current_costs.append(current_cost)

        if current_cost < personal_best_costs[i]:

            personal_best_costs[i] = current_cost

            personal_best_positions[i] = (
                positions[i].copy()
            )

            if current_cost < global_best_cost:

                global_best_cost = float(
                    current_cost
                )

                global_best_position = (
                    positions[i].copy()
                )

    optimization_history.append({
        "Iteration": iteration + 1,
        "Best_Fitness": -global_best_cost,
        "Mean_Fitness": -float(
            np.mean(current_costs)
        )
    })

    print(
        f"Iteration {iteration + 1}/{ITERATIONS}"
        f" | Best Fitness: {-global_best_cost:.4f}"
    )

# ================================================================
# 11. BEST PSO SOLUTION
# ================================================================

best_params, selected_features = decode_particle(
    global_best_position
)

best_cv_fitness = -global_best_cost

print("\n" + "=" * 75)
print("PSO OPTIMIZATION COMPLETED")
print("=" * 75)

print("\nBest parameters:")
print(best_params)

print("\nSelected features:")
print(selected_features)

print("\nBest CV fitness:", best_cv_fitness)

# ================================================================
# 12. TRAIN FINAL PSO MODEL
# ================================================================

pso_model = build_pipeline(best_params)

pso_model.fit(
    X_train[selected_features],
    y_train
)

pso_predictions = pso_model.predict(
    X_test[selected_features]
)

pso_probabilities = pso_model.predict_proba(
    X_test[selected_features]
)

# ================================================================
# 13. PERFORMANCE EVALUATION
# ================================================================

def evaluate(y_true, y_pred):

    return {
        "Accuracy": float(
            accuracy_score(y_true, y_pred)
        ),
        "Precision": float(
            precision_score(
                y_true,
                y_pred,
                average="weighted",
                zero_division=0
            )
        ),
        "Recall": float(
            recall_score(
                y_true,
                y_pred,
                average="weighted",
                zero_division=0
            )
        ),
        "F1 Score": float(
            f1_score(
                y_true,
                y_pred,
                average="weighted",
                zero_division=0
            )
        ),
        "Macro F1": float(
            f1_score(
                y_true,
                y_pred,
                average="macro",
                zero_division=0
            )
        ),
        "Balanced Accuracy": float(
            balanced_accuracy_score(
                y_true,
                y_pred
            )
        )
    }

pso_metrics = evaluate(
    y_test,
    pso_predictions
)

print("\nPSO MODEL PERFORMANCE")

for metric, value in pso_metrics.items():
    print(f"{metric}: {value:.4f}")

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        pso_predictions,
        labels=np.arange(n_classes),
        target_names=class_names,
        zero_division=0
    )
)

# ================================================================
# 14. BASELINE MODEL COMPARISON
# ================================================================

if baseline_bundle is not None:

    baseline_model = baseline_bundle["model"]

    baseline_features = baseline_bundle[
        "feature_columns"
    ]

    baseline_predictions = baseline_model.predict(
        X_test[baseline_features]
    )

else:

    baseline_model = build_pipeline({
        "n_estimators": 200,
        "max_depth": 5,
        "min_samples_split": 2,
        "min_samples_leaf": 2,
        "max_features": 1.0
    })

    baseline_model.fit(X_train, y_train)

    baseline_predictions = baseline_model.predict(
        X_test
    )

baseline_metrics = evaluate(
    y_test,
    baseline_predictions
)

# ================================================================
# 15. OPTIONAL AIS COMPARISON
# ================================================================

comparison_data = {
    "Baseline RF": baseline_metrics,
    "PSO-Optimized RF": pso_metrics
}

if AIS_PATH.exists():

    try:

        ais_bundle = joblib.load(AIS_PATH)

        ais_model = ais_bundle["model"]

        ais_features = ais_bundle[
            "selected_features"
        ]

        ais_predictions = ais_model.predict(
            X_test[ais_features]
        )

        ais_metrics = evaluate(
            y_test,
            ais_predictions
        )

        comparison_data["AIS-Optimized RF"] = (
            ais_metrics
        )

        print("\nAIS model loaded for comparison.")

    except Exception as error:

        print(
            "\nAIS comparison skipped:",
            str(error)
        )

comparison_df = pd.DataFrame.from_dict(
    comparison_data,
    orient="index"
)

comparison_df.to_csv(
    output("model_comparison.csv"),
    index_label="Model"
)

print("\nMODEL COMPARISON")
print(comparison_df.round(4))

# ================================================================
# 16. SAVE PSO MODEL
# ================================================================

pso_bundle = {
    "model": pso_model,
    "label_encoder": encoder,
    "selected_features": selected_features,
    "all_features": feature_columns,
    "best_parameters": best_params,
    "best_cv_fitness": best_cv_fitness,
    "risk_thresholds": {
        "low_max": q1,
        "medium_max": q2
    },
    "class_names": class_names,
    "random_seed": SEED
}

joblib.dump(
    pso_bundle,
    output("crime_risk_model.pkl")
)

print("\nSaved: pso_crime_risk_model.pkl")

# ================================================================
# 17. PSO ACCURACY GRAPH
# ================================================================

fig, ax = plt.subplots(figsize=(10, 6))

names = comparison_df.index.tolist()

accuracies = (
    comparison_df["Accuracy"].values * 100
)

colors = [
    "#2563eb",
    "#16a34a",
    "#f59e0b"
][:len(names)]

bars = ax.bar(
    names,
    accuracies,
    color=colors,
    width=0.55
)

for bar, value in zip(bars, accuracies):

    ax.text(
        bar.get_x() + bar.get_width() / 2,
        value + 1,
        f"{value:.2f}%",
        ha="center",
        fontweight="bold"
    )

ax.set_ylim(0, 110)
ax.set_ylabel("Accuracy (%)")

ax.set_title(
    "PSO Crime Risk Classification Accuracy",
    fontweight="bold",
    fontsize=15
)

ax.grid(axis="y", alpha=0.3)

save_plot("accuracy_graph.png")

# ================================================================
# 18. PSO CORRELATION HEATMAP
# ================================================================

heatmap_columns = list(dict.fromkeys(
    selected_features + [
        "2022",
        RATE_COL,
        CHARGE_COL
    ]
))

corr = df[heatmap_columns].corr(
    numeric_only=True
)

rename_columns = {
    "Crime_Growth_2020_2021": "Crime Growth",
    "Historical_Average_Crimes": "Average Crimes",
    "Historical_Crime_Difference": "Crime Difference",
    RATE_COL: "Crime Rate 2022",
    CHARGE_COL: "Chargesheeting 2022"
}

corr = corr.rename(
    index=rename_columns,
    columns=rename_columns
)

plt.figure(figsize=(11, 9))

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    vmin=-1,
    vmax=1,
    linewidths=0.5
)

plt.title(
    "PSO Selected Features - Correlation Heatmap",
    fontsize=15,
    fontweight="bold"
)

plt.xticks(rotation=45, ha="right")
plt.yticks(rotation=0)

save_plot("heatmap.png")

# ================================================================
# 19. PSO MODEL COMPARISON GRAPH
# ================================================================

metrics_to_plot = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1 Score"
]

fig, ax = plt.subplots(figsize=(12, 7))

x = np.arange(len(metrics_to_plot))

n_models = len(comparison_df)

width = 0.75 / n_models

for i, (name, row) in enumerate(
    comparison_df.iterrows()
):

    values = (
        row[metrics_to_plot].values.astype(float)
        * 100
    )

    positions = (
        x + (i - (n_models - 1) / 2) * width
    )

    bars = ax.bar(
        positions,
        values,
        width,
        label=name,
        color=colors[i]
    )

    for bar, value in zip(bars, values):

        ax.text(
            bar.get_x() + bar.get_width() / 2,
            value + 1,
            f"{value:.1f}",
            ha="center",
            fontsize=8
        )

ax.set_xticks(x)
ax.set_xticklabels(metrics_to_plot)

ax.set_ylabel("Score (%)")
ax.set_ylim(0, 110)

ax.set_title(
    "Baseline vs PSO vs AIS Model Performance",
    fontsize=15,
    fontweight="bold"
)

ax.legend()
ax.grid(axis="y", alpha=0.3)

save_plot("comparison_graph.png")

# ================================================================
# 20. PSO RESULT CSV
# ================================================================

result_df = pd.DataFrame({
    "State_UT": df.iloc[test_idx][
        "State/UT"
    ].values,

    "Actual_Risk": encoder.inverse_transform(
        y_test
    ),

    "PSO_Predicted_Risk":
        encoder.inverse_transform(
            pso_predictions
        ),

    "PSO_Confidence":
        np.max(pso_probabilities, axis=1),

    "Baseline_Predicted_Risk":
        encoder.inverse_transform(
            baseline_predictions
        )
})

result_df["PSO_Correct"] = (
    result_df["Actual_Risk"]
    == result_df["PSO_Predicted_Risk"]
)

result_df = result_df.sort_values(
    "State_UT"
).reset_index(drop=True)

result_df.to_csv(
    output("result.csv"),
    index=False
)

print("\nPSO TEST RESULTS")
print(result_df.to_string(index=False))

# ================================================================
# 21. PSO RESULT GRAPH
# ================================================================

risk_mapping = {
    "Low Risk": 0,
    "Medium Risk": 1,
    "High Risk": 2
}

actual_codes = (
    result_df["Actual_Risk"]
    .map(risk_mapping)
    .values
)

predicted_codes = (
    result_df["PSO_Predicted_Risk"]
    .map(risk_mapping)
    .values
)

x = np.arange(len(result_df))

fig, ax = plt.subplots(figsize=(14, 7))

ax.scatter(
    x - 0.13,
    actual_codes,
    color="black",
    s=110,
    marker="o",
    label="Actual"
)

ax.scatter(
    x + 0.13,
    predicted_codes,
    color="#16a34a",
    s=100,
    marker="s",
    label="PSO Predicted"
)

ax.set_xticks(x)

ax.set_xticklabels(
    result_df["State_UT"],
    rotation=60,
    ha="right"
)

ax.set_yticks([0, 1, 2])

ax.set_yticklabels([
    "Low Risk",
    "Medium Risk",
    "High Risk"
])

ax.set_ylim(-0.5, 2.5)

ax.set_ylabel("Crime Risk Category")

ax.set_title(
    "PSO Actual vs Predicted Crime Risk",
    fontsize=15,
    fontweight="bold"
)

ax.legend()
ax.grid(alpha=0.3)

save_plot("result_graph.png")

# ================================================================
# 22. PSO PREDICTION CSV
# ================================================================

# These are predictions for all states in the
# historical dataset, including training states.
# They are not future crime forecasts.

all_predictions = pso_model.predict(
    X[selected_features]
)

all_probabilities = pso_model.predict_proba(
    X[selected_features]
)

test_set = set(test_idx.tolist())

prediction_df = pd.DataFrame({
    "State_UT": df["State/UT"].values,

    "Crimes_2020": df["2020"].values,

    "Crimes_2021": df["2021"].values,

    "Crimes_2022": df["2022"].values,

    "Crime_Rate_2022": df[RATE_COL].values,

    "Chargesheeting_Rate_2022":
        df[CHARGE_COL].values,

    "Actual_Risk_2022":
        df["Crime_Risk"].values,

    "PSO_Predicted_Risk":
        encoder.inverse_transform(
            all_predictions
        ),

    "PSO_Confidence":
        np.max(all_probabilities, axis=1),

    "Data_Subset": [
        "Test" if i in test_set else "Train"
        for i in range(len(df))
    ]
})

prediction_df = prediction_df.sort_values(
    "Crime_Rate_2022",
    ascending=False
).reset_index(drop=True)

prediction_df.to_csv(
    output("prediction.csv"),
    index=False
)

print("\nPSO STATE-WISE PREDICTIONS")
print(prediction_df.to_string(index=False))

# ================================================================
# 23. PSO PREDICTION GRAPH
# ================================================================

plot_df = prediction_df.sort_values(
    "Crime_Rate_2022",
    ascending=True
)

risk_colors = {
    "Low Risk": "#16a34a",
    "Medium Risk": "#f59e0b",
    "High Risk": "#dc2626"
}

bar_colors = [
    risk_colors[risk]
    for risk in plot_df["PSO_Predicted_Risk"]
]

fig, ax = plt.subplots(figsize=(16, 11))

ax.barh(
    plot_df["State_UT"],
    plot_df["Crime_Rate_2022"],
    color=bar_colors
)

legend_items = [
    Patch(
        color=risk_colors["Low Risk"],
        label="PSO Low Risk"
    ),
    Patch(
        color=risk_colors["Medium Risk"],
        label="PSO Medium Risk"
    ),
    Patch(
        color=risk_colors["High Risk"],
        label="PSO High Risk"
    )
]

ax.legend(handles=legend_items)

ax.set_xlabel(
    "Reported IPC Crime Rate per Lakh Population (2022)"
)

ax.set_title(
    "PSO State-Wise Crime Risk Classification",
    fontsize=16,
    fontweight="bold"
)

save_plot("prediction_graph.png")

# ================================================================
# 24. PSO CONFUSION MATRIX
# ================================================================

cm = confusion_matrix(
    y_test,
    pso_predictions,
    labels=np.arange(n_classes)
)

plt.figure(figsize=(8, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=class_names,
    yticklabels=class_names
)

plt.title(
    "PSO-Optimized Random Forest Confusion Matrix",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Predicted")
plt.ylabel("Actual")

save_plot("confusion_matrix.png")

# ================================================================
# 25. PSO OPTIMIZATION GRAPH
# ================================================================

history_df = pd.DataFrame(
    optimization_history
)

history_df.to_csv(
    output("optimization_history.csv"),
    index=False
)

fig, ax = plt.subplots(figsize=(10, 6))

ax.plot(
    history_df["Iteration"],
    history_df["Best_Fitness"],
    marker="o",
    linewidth=2.5,
    label="Global Best Fitness"
)

ax.plot(
    history_df["Iteration"],
    history_df["Mean_Fitness"],
    marker="s",
    linestyle="--",
    label="Mean Swarm Fitness"
)

ax.set_xlabel("PSO Iteration")
ax.set_ylabel("Cross-Validated Fitness")

ax.set_title(
    "PSO Optimization Convergence",
    fontsize=15,
    fontweight="bold"
)

ax.legend()
ax.grid(alpha=0.3)

save_plot("optimization_graph.png")

# ================================================================
# 26. PSO FEATURE IMPORTANCE
# ================================================================

importance = pso_model.named_steps[
    "classifier"
].feature_importances_

importance_df = pd.DataFrame({
    "Feature": selected_features,
    "Importance": importance
}).sort_values(
    "Importance",
    ascending=False
)

importance_df.to_csv(
    output("feature_importance.csv"),
    index=False
)

plt.figure(figsize=(11, 6))

sns.barplot(
    data=importance_df,
    x="Importance",
    y="Feature",
    color="#16a34a"
)

plt.title(
    "PSO Selected Feature Importance",
    fontsize=15,
    fontweight="bold"
)

save_plot("feature_importance.png")

# ================================================================
# 27. CLASSIFICATION REPORT CSV
# ================================================================

report = classification_report(
    y_test,
    pso_predictions,
    labels=np.arange(n_classes),
    target_names=class_names,
    output_dict=True,
    zero_division=0
)

report_df = pd.DataFrame(report).transpose()

report_df.to_csv(
    output("classification_report.csv"),
    index_label="Category"
)

print("\nSaved: pso_classification_report.csv")

# ================================================================
# 28. PSO CONFIGURATION YAML
# ================================================================

config = {
    "project": {
        "name": "PSO-Based Crime Risk Assessment",
        "dataset": str(DATASET_PATH),
        "output_directory": str(PROJECT_DIR),
        "algorithm": "Particle Swarm Optimization",
        "random_seed": SEED
    },

    "pso": {
        "particles": N_PARTICLES,
        "iterations": ITERATIONS,
        "inertia": INERTIA,
        "cognitive_coefficient": COGNITIVE,
        "social_coefficient": SOCIAL,
        "fitness_metric": "macro_f1",
        "cross_validation_folds": 3,
        "feature_selection": True
    },

    "optimized_model": {
        "type": "RandomForestClassifier",
        "parameters": best_params,
        "selected_features": selected_features
    },

    "risk_thresholds": {
        "low_max": q1,
        "medium_max": q2
    },

    "performance": pso_metrics,

    "limitations": [
        "Small state-level dataset",
        "Risk categories based on 2022 crime rates",
        "2022 crime outcomes excluded from predictors",
        "Single held-out test split",
        "External validation required"
    ]
}

with open(
    output("config.yaml"),
    "w",
    encoding="utf-8"
) as f:

    yaml.safe_dump(
        config,
        f,
        sort_keys=False,
        allow_unicode=True
    )

print("Saved: pso_config.yaml")

# ================================================================
# 29. PSO MODEL INFORMATION JSON
# ================================================================

model_info = {
    "project_name": "PSO-Based Crime Risk Assessment",

    "created_at": datetime.now().isoformat(),

    "optimization_algorithm": (
        "Particle Swarm Optimization"
    ),

    "base_classifier": "Random Forest",

    "dataset": {
        "total_records": int(len(df)),
        "training_records": int(len(X_train)),
        "testing_records": int(len(X_test)),
        "selected_features": selected_features,
        "risk_classes": class_names
    },

    "optimization": {
        "particles": N_PARTICLES,
        "iterations": ITERATIONS,
        "best_cv_fitness": float(best_cv_fitness),
        "best_parameters": best_params,
        "unique_fitness_evaluations": len(fitness_cache)
    },

    "pso_performance": pso_metrics,

    "baseline_performance": baseline_metrics,

    "confusion_matrix": cm.tolist(),

    "files": {
        "model": "pso_crime_risk_model.pkl",
        "config": "pso_config.yaml",
        "metadata": "pso_model_info.json",
        "results": "pso_result.csv",
        "predictions": "pso_prediction.csv",
        "accuracy_graph": "pso_accuracy_graph.png",
        "heatmap": "pso_heatmap.png",
        "comparison_graph": "pso_comparison_graph.png",
        "optimization_graph": "pso_optimization_graph.png"
    }
}

with open(
    output("model_info.json"),
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        model_info,
        f,
        indent=4,
        ensure_ascii=False
    )

print("Saved: pso_model_info.json")

# ================================================================
# 30. VERIFY ALL PSO OUTPUTS
# ================================================================

expected_files = [
    "crime_risk_model.pkl",
    "config.yaml",
    "model_info.json",
    "accuracy_graph.png",
    "heatmap.png",
    "comparison_graph.png",
    "result.csv",
    "result_graph.png",
    "prediction.csv",
    "prediction_graph.png",
    "confusion_matrix.png",
    "optimization_graph.png",
    "optimization_history.csv",
    "classification_report.csv",
    "model_comparison.csv",
    "feature_importance.csv",
    "feature_importance.png"
]

print("\n" + "=" * 75)
print("PSO OUTPUT VERIFICATION")
print("=" * 75)

for filename in expected_files:

    path = output(filename)

    if path.exists():

        size_kb = path.stat().st_size / 1024

        print(
            f"[SUCCESS] {path.name}"
            f" ({size_kb:.2f} KB)"
        )

    else:

        print(f"[MISSING] {path.name}")

# ================================================================
# 31. FINAL SUMMARY
# ================================================================

print("\n" + "=" * 75)
print("PSO PROJECT COMPLETED")
print("=" * 75)

print("\nBest PSO Parameters:")
print(best_params)

print("\nSelected Features:")
print(selected_features)

print("\nBest Cross-Validation Fitness:")
print(round(best_cv_fitness, 4))

print("\nBaseline Accuracy:")
print(
    round(
        baseline_metrics["Accuracy"] * 100,
        2
    ),
    "%"
)

print("\nPSO Accuracy:")
print(
    round(
        pso_metrics["Accuracy"] * 100,
        2
    ),
    "%"
)

print("\nBaseline Macro F1:")
print(
    round(
        baseline_metrics["Macro F1"],
        4
    )
)

print("\nPSO Macro F1:")
print(
    round(
        pso_metrics["Macro F1"],
        4
    )
)

print("\nModel Comparison:")
print(comparison_df.round(4))

print("\nOutput Directory:")
print(PROJECT_DIR)

print("\nAll PSO outputs generated successfully.")

print(
    "\nResearch caution: Results are exploratory. "
    "Only 36 state/UT observations are available, "
    "and a single test split cannot establish "
    "reliable predictive performance."
)
```
